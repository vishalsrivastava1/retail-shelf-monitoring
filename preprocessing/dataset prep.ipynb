```python
import cv2
import os
import random

# Paths
IMAGE_DIR = "data/data_subset/image_subset/train"
LABEL_DIR = "data/data_subset/label_subset/train"

# Random image
images = os.listdir(IMAGE_DIR)
sample_image = random.choice(images)

# Paths
img_path = os.path.join(IMAGE_DIR, sample_image)

label_path = os.path.join(
    LABEL_DIR,
    sample_image.replace(".jpg", ".txt")
)

# Read image
img = cv2.imread(img_path)

# Get image dimensions
h, w, _ = img.shape

# Read labels
with open(label_path, "r") as f:
    lines = f.readlines()

# Draw bounding boxes
for line in lines:

    cls, xc, yc, bw, bh = map(float, line.strip().split())

    # Convert YOLO format → pixel coordinates
    x1 = int((xc - bw / 2) * w)
    y1 = int((yc - bh / 2) * h)
    x2 = int((xc + bw / 2) * w)
    y2 = int((yc + bh / 2) * h)

    # Draw rectangle
    cv2.rectangle(
        img,
        (x1, y1),
        (x2, y2),
        (0, 255, 0),
        2
    )

# Resize image for viewing
img_resized = cv2.resize(img, (1200, 800))

# Show image
cv2.imshow("Bounding Boxes", img_resized)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
